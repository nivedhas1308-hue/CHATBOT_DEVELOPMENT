```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful education chatbot. Explain topics in simple words."
    }
]

def ask_chatbot(question):
    messages.append({"role": "user", "content": question})

    response = ollama.chat(
        model=MODEL,
        messages=messages
    )

    answer = response["message"]["content"]
    messages.append({"role": "assistant", "content": answer})

    return answer

print(ask_chatbot("What is Python?"))
print("\nFollow-up question:")
print(ask_chatbot("Explain it in simple words."))
```

```python
from datetime import datetime

history_file = DATASET / "notebook_chat_history.csv"

question = "What is machine learning?"
answer = ask_chatbot(question)

file_exists = history_file.exists()

with open(history_file, "a", newline="", encoding="utf-8") as file:
    writer = __import__("csv").writer(file)

    if not file_exists:
        writer.writerow(["timestamp", "user_question", "chatbot_response"])

    writer.writerow([
        datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        question,
        answer
    ])

print("Question:", question)
print("Response:", answer)
print("Saved to:", history_file)
```
